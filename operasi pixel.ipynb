# Import Library

```python
import cv2
import matplotlib.pyplot as plt
```

# Membaca Gambar

```python
img = cv2.imread('blacklight.jpeg')
```

# Konversi BGR ke RGB

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

# Konversi ke grayscale

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

# Meningkatkan kecerahan

```python
bright = cv2.convertScaleAbs(gray, alpha=1.0, beta=50)
```

# Meningkatkan kontras

```python
contrast = cv2.convertScaleAbs(gray, alpha=2.0, beta=0)
```

# Meningkatkan kecerahan dan kontras

```python
bright_contrast = cv2.convertScaleAbs(gray, alpha=2.0, beta=50)
```

# Menampilkan hasil

```python
plt.figure(figsize=(8,12))

plt.subplot(5,1,1)
plt.imshow(img_rgb)
plt.title("Gambar Asli")
plt.axis('off')

plt.subplot(5,1,2)
plt.imshow(gray, cmap='gray')
plt.title("Gambar Gray")
plt.axis('off')

plt.subplot(5,1,3)
plt.imshow(bright, cmap='gray')
plt.title("Gambar Gray yang Dipercerah")
plt.axis('off')

plt.subplot(5,1,4)
plt.imshow(contrast, cmap='gray')
plt.title("Gambar Gray yang Diperkontras")
plt.axis('off')

plt.subplot(5,1,5)
plt.imshow(bright_contrast, cmap='gray')
plt.title("Gambar Gray yang Dipercerah dan Diperkontras")
plt.axis('off')

plt.tight_layout()
plt.show()
```
