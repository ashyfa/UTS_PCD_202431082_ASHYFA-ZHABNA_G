# Import Library

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# Membaca Gambar

```python
# Membaca gambar
img = cv2.imread("gambar nama uts pcd.jpeg")

# Konversi dari BGR ke RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

# MEMISAHKAN CHANNEL WARNA

```python
# Channel Merah
R = img_rgb[:, :, 0]

# Channel Hijau
G = img_rgb[:, :, 1]

# Channel Biru
B = img_rgb[:, :, 2]
```

# MENAMPILKAN HASIL DETEKSI WARNA

```python
plt.figure(figsize=(12,8))

# Gambar Asli
plt.subplot(2,2,1)
plt.imshow(img_rgb)
plt.title("Citra Asli")
plt.axis("off")

# Warna Biru
plt.subplot(2,2,2)
plt.imshow(B, cmap='gray')
plt.title("Deteksi Warna Biru")
plt.axis("off")

# Warna Merah
plt.subplot(2,2,3)
plt.imshow(R, cmap='gray')
plt.title("Deteksi Warna Merah")
plt.axis("off")

# Warna Hijau
plt.subplot(2,2,4)
plt.imshow(G, cmap='gray')
plt.title("Deteksi Warna Hijau")
plt.axis("off")

plt.tight_layout()
plt.show()
```

# MEMBUAT HISTOGRAM RGB

```python
plt.figure(figsize=(10,5))

# Histogram Merah
plt.hist(R.ravel(), bins=256, color='red', alpha=0.5, label='Merah')

# Histogram Hijau
plt.hist(G.ravel(), bins=256, color='green', alpha=0.5, label='Hijau')

# Histogram Biru
plt.hist(B.ravel(), bins=256, color='blue', alpha=0.5, label='Biru')

plt.title("Histogram RGB")
plt.xlabel("Intensitas Pixel")
plt.ylabel("Jumlah Pixel")
plt.legend()

plt.show()
```
